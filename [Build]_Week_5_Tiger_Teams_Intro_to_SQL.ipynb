# 🔍 **Introduction: Why SQL?**

**Structured Query Language (SQL)** is a programming language used to **store, query, and manage data** in databases.

If Pandas is like Excel in Python, then SQL is like the **language that databases speak**.
While Pandas works with in-memory DataFrames, SQL works with tables stored in databases — but both share similar ideas:

*   Rows = records
*   Columns = fields

You can filter, sort, group, and join data — just like with Pandas!

In this assignment, you’ll:

1. Learn how to use **SQL in Google Colab**.

2. Compare **SQL queries** with familiar **Pandas operations**.

3. Write your own simple SQL queries!

The general syntax of a SQL query follows this form, though you may not need all of these arguments in every query you write:
```python
SELECT column1, column2, aggregate_function(column3), ...
FROM table_name
WHERE condition
GROUP BY column_name(s)
HAVING condition
ORDER BY column_name(s) [ASC | DESC];

## **Set Up**

````python
# Run this cell to import pandas and sqlite
import sqlite3
import pandas as pd

# Create a sample dataset
data = {
    "name": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "age": [24, 27, 22, 32, 29],
    "city": ["New York", "Los Angeles", "New York", "Chicago", "Chicago"],
    "score": [88, 92, 85, 79, 95]
}

df = pd.DataFrame(data)

# Create an in-memory SQLite database and load the dataframe
conn = sqlite3.connect(":memory:")
df.to_sql("students", conn, index=False, if_exists="replace")

# View the data
pd.read_sql("SELECT * FROM students", conn)
````

## **Step 1: Your First SQL Query**

In Pandas, you might filter rows like this to show only the rows in the DataFrame where the score is greater than 85:

```python
df[df["score"] > 85]

In SQL, you write:

```python
SELECT * FROM students WHERE score > 85;

This *selects* **all columns** (using *) *from* the **students** table *where* the **score > 85**.  

We will use the following syntax to display SQL queries using the read_sql method in the pandas library. Try running the cell below.

````python
pd.read_sql("SELECT * FROM students WHERE score > 85;", conn)
````
